```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
plt.rcParams['svg.fonttype'] = 'none'
```

```python
df = pd.read_csv('../pre-calculated_data/number_ARGs_per_sample.csv')
df
```

```python
custom_palette = {
    'Nanjing': '#1b9e77',
    'Shanghai': '#e6ab02',
    'SMAG': '#7570b3'
}

fig, ax = plt.subplots(figsize=(4, 5))
sns.boxplot(data=df, x="City", y="n", color='white', hue="City", 
             palette=custom_palette,boxprops=dict(alpha=0.5),
            showfliers=False, ax=ax, width=0.6, order=["Nanjing", "Shanghai"])

sns.stripplot(data=df, x="City", y="n", hue="City", 
             palette=custom_palette, ax=ax,size=5, jitter=True, order=["Nanjing", "Shanghai"])


ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_ylabel('ARGs per sample', fontsize=12)
ax.set_xlabel('')
plt.savefig('figure7a.svg')
```
